```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import librosa
import librosa.display
```

```python
dataset_path = "../torgo"

print(dataset_path)

import os

print(os.path.exists(dataset_path))
```

```python
df = pd.read_csv("../torgo_metadata.csv")

print(df.shape)
df.head()
```

```python
control_file = df[df["condition"] == "Control"].iloc[0]["file"]

print(control_file)
```

```python
audio, sr = librosa.load(control_file, sr=None)
```

```python
print("Type:", type(audio))
print("Sampling rate:", sr)
print("Number of samples:", len(audio))
print("Duration:", len(audio) / sr, "seconds")
print("Minimum amplitude:", audio.min())
print("Maximum amplitude:", audio.max())
```

```python
plt.figure(figsize=(12, 4))

librosa.display.waveshow(audio, sr=sr)

plt.title("Control Speech - Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.show()
```

```python
dysarthria_file = df[df["condition"] == "Dysarthria"].iloc[0]["file"]

print(dysarthria_file)
```

```python
dys_audio, dys_sr = librosa.load(
    dysarthria_file,
    sr=None
)

print("Sampling rate:", dys_sr)
print("Number of samples:", len(dys_audio))
print("Duration:", len(dys_audio) / dys_sr, "seconds")
print("Minimum amplitude:", audio.min())
print("Maximum amplitude:", audio.max())
```

```python
plt.figure(figsize=(12, 4))

librosa.display.waveshow(dys_audio, sr=dys_sr)

plt.title("Dysarthric Speech - Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")

plt.show()
```

```python
plt.figure(figsize=(14, 7))

plt.subplot(2, 1, 1)
librosa.display.waveshow(audio, sr=sr)
plt.title("Control Speech - Waveform")
plt.ylabel("Amplitude")

plt.subplot(2, 1, 2)
librosa.display.waveshow(dys_audio, sr=dys_sr)
plt.title("Dysarthric Speech - Waveform")
plt.ylabel("Amplitude")
plt.xlabel("Time (seconds)")

plt.tight_layout()
plt.show()
```
